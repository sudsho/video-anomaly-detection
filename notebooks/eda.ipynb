# UCSD Ped2 EDA

Quick look at the dataset: count of clips, frame distribution, sample frames, anomaly label distribution per video.

```python
import os, glob
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
ROOT = '../data/raw/UCSD_Anomaly_Dataset/UCSDped2'
train_dirs = sorted(glob.glob(os.path.join(ROOT, 'Train', 'Train*')))
test_dirs  = sorted(glob.glob(os.path.join(ROOT, 'Test',  'Test*')))
print('train videos:', len(train_dirs), 'test videos:', len(test_dirs))
```

```python
def n_frames(d):
    return len(glob.glob(os.path.join(d, '*.tif'))) + len(glob.glob(os.path.join(d, '*.jpg')))

lens = [n_frames(d) for d in train_dirs + test_dirs]
print('min/median/max frames per clip:', np.min(lens), np.median(lens), np.max(lens))
```

```python
if test_dirs:
    frames = sorted(glob.glob(os.path.join(test_dirs[0], '*.tif')))[:8]
    fig, axes = plt.subplots(2, 4, figsize=(12, 6))
    for ax, fp in zip(axes.flat, frames):
        img = cv2.cvtColor(cv2.imread(fp), cv2.COLOR_BGR2RGB)
        ax.imshow(img); ax.axis('off')
    plt.suptitle('Sample frames from Test001')
    plt.show()
```

```python
# anomaly distribution per test video
from src.labels import load_ucsdped2_frame_labels
labels = load_ucsdped2_frame_labels(os.path.join(ROOT, 'Test'))
ratios = {k: float(np.mean(v)) for k, v in labels.items()}
print(sorted(ratios.items(), key=lambda kv: -kv[1])[:5])
```
